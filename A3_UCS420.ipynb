**Q.6 Create a sample dataset (employees.csv) containing information about employees in a company.**

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(data)

employees.to_csv("employees.csv", index=False)

print(employees)
```

```python
import pandas as pd

df = pd.read_csv("employees.csv")

# 6(a) Display the shape (number of rows and columns) of the DataFrame.
print(df.shape)

# 6(b) Display a summary of the DataFrame including data types and non-null counts for each column.
df.info()

# 6(c) Generate descriptive statistics.
print(df.describe())

# 6(d) Display the first 5 rows and last 3 rows of the dataset.
print(df.head(5))
print(df.tail(3))

# 6(e)(i) Find the average salary of employees.
print(df["Salary"].mean())

# 6(e)(ii) Find the total bonus paid to all employees.
print(df["Bonus"].sum())

# 6(e)(iii) Find the youngest employee's age.
print(df["Age"].min())

# 6(e)(iv) Find the highest performance rating.
print(df["Rating"].max())

# 6(f) Sort the DataFrame by the Salary column in descending order.
print(df.sort_values(by="Salary", ascending=False))

# 6(g) Add a new column that categorizes employees based on their performance rating.
df["Performance"] = df["Rating"].apply(
    lambda x: "Excellent" if x >= 4.5 else "Good" if x >= 4.0 else "Average"
)
print(df)

# 6(h) Identify missing values in the DataFrame.
print(df.isnull().sum())

# 6(i) Rename the Employee_ID column to ID.
df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print(df)

# 6(j)(i) Find all employees who have more than 5 years of experience.
print(df[df["Years_of_Experience"] > 5])

# 6(j)(ii) Find all employees who belong to the IT department.
print(df[df["Department"] == "IT"])

# 6(k) Add a new column Tax, which deducts 10% of the Salary.
df["Tax"] = df["Salary"] * 0.10
print(df)

# 6(l) Save the modified DataFrame with added columns to a new CSV file.
df.to_csv("modified_employees.csv", index=False)
```
