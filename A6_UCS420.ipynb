Q2. Matrix representing daily steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a) Total steps
total_steps = np.sum(steps)

# b) Mean steps
mean_steps = np.mean(steps)

# c) Maximum and minimum
maximum = np.max(steps)
minimum = np.min(steps)

# d) Total steps for each day
daily_total = np.sum(steps, axis=0)

# e) Total steps for each user
user_total = np.sum(steps, axis=1)

# f) Position of maximum value
max_position = np.unravel_index(
    np.argmax(steps), steps.shape
)
```

Q3. NumPy slicing, copying, reshaping and flattening

```python
# a) Create original array
original = np.array([1, 2, 3, 4, 5, 6])

# b) Slice from index 1 to 4
subset = original[1:5]

# c) Modify subset
subset[0] = 999

print(original)
print(subset)

# d) Create a copy and modify it
copied = original[1:5].copy()

copied[0] = 500

print(original)
print(copied)

# e) Create 3 × 4 matrix
matrix = np.arange(1, 13).reshape(3, 4)

# f) Extract elements
first_row = matrix[0]
last_row = matrix[-1]
second_column = matrix[:, 1]
selected_elements = matrix[0:2, 1:3]

print(first_row)
print(last_row)
print(second_column)
print(selected_elements)

# g) flatten() and ravel()
flattened = matrix.flatten()
raveled = matrix.ravel()

print(flattened)
print(raveled)

# h) Modify ravel()
raveled[0] = 100

print(matrix)
print(raveled)

# i) Modify flatten()
flattened[1] = 200

print(matrix)
print(flattened)

# j) Shape, dimensions, size and dtype
print(matrix.shape)
print(matrix.ndim)
print(matrix.size)
print(matrix.dtype)
```

Q4. Cognitive-assistive system

```python
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# a) Shape and dimensions
print(X.shape)
print(X.ndim)

# b) Transpose
XT = X.T
print(XT)

# c) Matrix product
XtX = X.T @ X
print(XtX)

# d) Matrix inverse
inverse = np.linalg.inv(XtX)
print(inverse)

# e) Ordinary Least Squares
beta = np.linalg.inv(X.T @ X) @ X.T @ y
print(beta)

# g) Predicted assistance score
new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta
print(predicted_score)
```
