iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_prime(n):
 if n<2:
   return False
 for i in range(2,int(n*(1/2)+1)):
   if n%i==0:
    return False

 return True
prime_numbers=[x for x in numbers if is_prime(x)]
print("\nPrime numbers")
print(prime_numbers)
print("Number of prime numbers:",len(prime_numbers))
```

v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
frequency={}
for x in numbers:
  frequency[x]=frequency.get(x,0)+1
  most_freq=max(frequency,key=frequency.get)
print("Most frequenct number is:",most_freq)
print("It occurs:",frequency[most_freq],"times")
```

```python
# Q4: Set operations using the 8 digits of the roll number

# Enter the same 8 digits from Q1 (before multiplying by 10)
roll_digits = [1, 2, 3, 4, 5, 6, 7, 8]   # <-- replace with your digits

# Create sets A and B
A = {digit * 7 for digit in roll_digits}
B = {digit * 9 for digit in roll_digits}

print("Set A =", A)
print("Set B =", B)

# vi. Union
print("Union of A and B =", A.union(B))

# vii. Intersection
print("Intersection of A and B =", A.intersection(B))

# viii. A - B and B - A separately
print("A - B =", A.difference(B))
print("B - A =", B.difference(A))

# Difference() gives values present in one set but not the other;
# symmetric_difference() gives all values that are in either set but not in both.

# ix. Symmetric difference
print("Symmetric Difference =", A.symmetric_difference(B))

# x. Subset and Superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. Input X and discard it from A
X = int(input("Enter a value X to remove from set A: "))

A.discard(X)

print("Set A after discard =", A)
```

```python
# Q5 - Dictionary Operations

# Original dictionary
my_dict = {
    "name": "Zaira",
    "roll_no": "1024160093",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

# i. Rename "city" to "location" using pop()
my_dict["location"] = my_dict.pop("city")

print("i. After renaming city:")
print(my_dict)


# ii. Add a new key "cgpa"
my_dict["cgpa"] = 8.5

print("\nii. After adding CGPA:")
print(my_dict)


# iii. Increase age by 1
my_dict["age"] += 1

print("\niii. After increasing age:")
print(my_dict)


# iv. Delete "branch" using pop() and del
# Create two separate copies
dict_pop = my_dict.copy()
dict_del = my_dict.copy()

# Using pop()
removed_value = dict_pop.pop("branch")

print("\niv. Using pop():")
print(dict_pop)
print("Returned value:", removed_value)

# Using del
del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value,
# whereas del removes the key but does not return its value.


# v. Iterate using items()
print("\nv. Key-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)


# vi. Check whether "email" exists
print("\nvi. Checking email:")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email does not exist in the dictionary.")


# vii. Create friend's dictionary
friend_dict = {
    "name": "Muskan",
    "roll_no": "1024160074",
    "branch": "ECE",
    "age": 21,
    "city": "Mumbai"
}

# Merge using ** syntax
merged_dict = {**my_dict, **friend_dict}

print("\nvii. Merged dictionary:")
print(merged_dict)

# If both dictionaries have the same key,
# the value from the second dictionary wins.


# viii. Dictionary comprehension
# Create a dictionary containing only string values
string_dict = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nviii. Dictionary with only string values:")
print(string_dict)
```
