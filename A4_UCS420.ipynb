Q2: Generate and Score a Hypothesis

```python
def score_query(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words & keywords)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    return pd.DataFrame(results).sort_values("score", ascending=False)


query = input("Enter your query: ")
print(score_query(query, df))
```

Q6: Modify Q2 scoring function to handle ties

```python
def score_query_with_ties(query, df):
    query_words = set(query.lower().split())
    results = []

    for _, row in df.iterrows():
        keywords = set(row["keywords"].lower().split())
        score = len(query_words & keywords)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    if not results:
        print("No matching entries found.")
        return

    result = pd.DataFrame(results)
    highest_score = result["score"].max()

    print(result[result["score"] == highest_score])


query = input("Enter a query for Q6: ")
score_query_with_ties(query, df)
```
